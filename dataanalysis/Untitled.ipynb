```python
import pandas as pd

# Load Excel file
file_path = "iitmindia.xlsx"
df = pd.read_excel(file_path)

# Basic overview
print("Shape:", df.shape)
print("\nColumns:", df.columns.tolist())
print("\nData Types:\n", df.dtypes)

# First rows
print("\nHead:\n", df.head())

# Missing values
print("\nMissing Values:\n", df.isnull().sum().sort_values(ascending=False))

# Basic stats (numeric columns)
print("\nDescribe:\n", df.describe(include="all"))

# Duplicate rows
print("\nDuplicate rows:", df.duplicated().sum())

# Unique values per column (top 5 columns only to avoid overload)
print("\nUnique counts (sample):")
for col in df.columns[:10]:
    print(col, "->", df[col].nunique())
```

```python
import pandas as pd

# Load Excel file
file_path = "iitmindia.xlsx"
df = pd.read_excel(file_path)

# Clean basic placeholders
df = df.replace(["-", " ", ""], pd.NA)

# Remove entries with no email AND no mobile
df = df.dropna(subset=["email", "mobile"], how="any")

# Optional: remove rows where email/mobile are empty after stripping
df = df[df["email"].astype(str).str.strip().ne("") & df["mobile"].astype(str).str.strip().ne("")]

# Reset index
df = df.reset_index(drop=True)

print("Shape after cleaning:", df.shape)
print(df.head())
```

```python
# Export to Excel
output_path = "output.xlsx"
df.to_excel(output_path, index=False)

print("Saved to:", output_path)
print("Shape:", df.shape)
```

```python
import pandas as pd

# Load Excel file
file_path = "iitmindia.xlsx"
df = pd.read_excel(file_path)

# Clean placeholders
df = df.replace(["-", " ", ""], pd.NA)

# Remove entries with no email AND no mobile
df = df.dropna(subset=["email", "mobile"], how="any")

# Clean whitespace
df["state"] = df["state"].astype(str).str.strip()

# Count by state
state_counts = df["state"].value_counts(dropna=False)

print(state_counts)

# Optional: save result
state_counts.to_excel("state_counts.xlsx", header=["count"])
```
