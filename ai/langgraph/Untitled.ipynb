```python
# ============================================================
# SIMPLE QWEN 2.5 7B CHATBOT
# Flask + Ollama
# JupyterLab
# Port: 8080
# ============================================================

import os
import sys
import subprocess
import threading
import time

# ------------------------------------------------------------
# Install dependencies
# ------------------------------------------------------------

subprocess.check_call([
    sys.executable,
    "-m",
    "pip",
    "install",
    "-q",
    "flask",
    "requests"
])

# ------------------------------------------------------------
# Create folders
# ------------------------------------------------------------

os.makedirs("qwen_chatbot/templates", exist_ok=True)


# ------------------------------------------------------------
# Flask application
# ------------------------------------------------------------

app_code = r'''
from flask import Flask, render_template, request, jsonify
import requests

app = Flask(__name__)

# Your actual installed Ollama model
MODEL = "qwen2.5:7b"

# Ollama API
OLLAMA_URL = "http://localhost:11434/api/chat"


@app.route("/")
def home():
    return render_template("index.html")


@app.route("/chat", methods=["POST"])
def chat():

    data = request.get_json()

    user_message = data.get("message", "").strip()

    if not user_message:
        return jsonify({
            "response": "Please enter a message."
        })


    try:

        ollama_response = requests.post(
            OLLAMA_URL,

            json={
                "model": MODEL,

                "messages": [
                    {
                        "role": "user",
                        "content": user_message
                    }
                ],

                "stream": False
            },

            timeout=300
        )


        ollama_response.raise_for_status()

        result = ollama_response.json()

        answer = result["message"]["content"]


        return jsonify({
            "response": answer
        })


    except Exception as e:

        return jsonify({
            "response": f"Ollama error: {str(e)}"
        }), 500


if __name__ == "__main__":

    app.run(
        host="0.0.0.0",
        port=8080,
        debug=False,
        use_reloader=False
    )
'''


with open("qwen_chatbot/app.py", "w", encoding="utf-8") as f:
    f.write(app_code)


# ------------------------------------------------------------
# Chat interface
# ------------------------------------------------------------

html_code = r'''
<!DOCTYPE html>

<html>

<head>

    <meta charset="UTF-8">

    <title>Qwen Chatbot</title>


    <style>

        * {
            box-sizing: border-box;
        }


        body {

            margin: 0;

            font-family: Arial, sans-serif;

            background: #f5f5f5;
        }


        .chat {

            width: 750px;

            max-width: 95%;

            margin: 40px auto;

            background: white;

            border-radius: 12px;

            overflow: hidden;

            box-shadow:
                0 4px 20px
                rgba(0,0,0,.12);
        }


        .header {

            background: #202020;

            color: white;

            padding: 20px;

            font-size: 22px;

            font-weight: bold;
        }


        .model {

            font-size: 13px;

            color: #aaa;

            margin-top: 5px;

            font-weight: normal;
        }


        #messages {

            height: 500px;

            overflow-y: auto;

            padding: 20px;
        }


        .message {

            padding: 12px 16px;

            margin-bottom: 15px;

            border-radius: 10px;

            line-height: 1.5;

            white-space: pre-wrap;

            word-wrap: break-word;
        }


        .user {

            background: #dbeafe;

            margin-left: 100px;
        }


        .assistant {

            background: #eeeeee;

            margin-right: 100px;
        }


        .input-area {

            display: flex;

            padding: 15px;

            border-top: 1px solid #ddd;
        }


        #input {

            flex: 1;

            padding: 13px;

            border: 1px solid #ccc;

            border-radius: 7px;

            font-size: 16px;

            outline: none;
        }


        #input:focus {

            border-color: #555;
        }


        #send {

            margin-left: 10px;

            padding: 13px 24px;

            border: none;

            border-radius: 7px;

            background: #202020;

            color: white;

            font-size: 16px;

            cursor: pointer;
        }


        #send:hover {

            background: #444;
        }


        #send:disabled {

            background: #999;

            cursor: not-allowed;
        }

    </style>

</head>


<body>


<div class="chat">


    <div class="header">

        Qwen Chatbot

        <div class="model">

            qwen2.5:7b • Ollama • Local

        </div>

    </div>


    <div id="messages"></div>


    <div class="input-area">

        <input
            id="input"
            type="text"
            placeholder="Type a message..."
            autocomplete="off"
        >

        <button id="send">
            Send
        </button>

    </div>


</div>


<script>


const input =
    document.getElementById("input");


const send =
    document.getElementById("send");


const messages =
    document.getElementById("messages");



function addMessage(
    name,
    text,
    className
) {

    const div =
        document.createElement("div");


    div.className =
        "message " + className;


    div.textContent =
        name + ": " + text;


    messages.appendChild(div);


    messages.scrollTop =
        messages.scrollHeight;


    return div;
}



async function sendMessage() {


    const text =
        input.value.trim();


    if (!text) {
        return;
    }


    addMessage(
        "You",
        text,
        "user"
    );


    input.value = "";


    input.disabled = true;

    send.disabled = true;


    const thinking =
        addMessage(
            "Qwen",
            "Thinking...",
            "assistant"
        );


    try {


        const response =
            await fetch(
                "/chat",
                {

                    method: "POST",

                    headers: {
                        "Content-Type":
                            "application/json"
                    },

                    body:
                        JSON.stringify({
                            message: text
                        })

                }
            );


        const data =
            await response.json();


        thinking.remove();


        addMessage(
            "Qwen",
            data.response,
            "assistant"
        );


    } catch (error) {


        thinking.remove();


        addMessage(
            "Qwen",
            "Connection error: " +
            error.message,
            "assistant"
        );

    }


    input.disabled = false;

    send.disabled = false;

    input.focus();

}



send.addEventListener(
    "click",
    sendMessage
);



input.addEventListener(
    "keydown",
    function(event) {

        if (event.key === "Enter") {

            sendMessage();

        }

    }
);


input.focus();

</script>


</body>

</html>
'''


with open(
    "qwen_chatbot/templates/index.html",
    "w",
    encoding="utf-8"
) as f:

    f.write(html_code)


# ------------------------------------------------------------
# Start Flask
# ------------------------------------------------------------

os.chdir("qwen_chatbot")

from app import app


def run_server():

    app.run(
        host="0.0.0.0",
        port=8080,
        debug=False,
        use_reloader=False
    )


thread = threading.Thread(
    target=run_server,
    daemon=True
)

thread.start()


time.sleep(2)


print()
print("=" * 60)
print("QWEN CHATBOT STARTED")
print("=" * 60)
print()
print("Model : qwen2.5:7b")
print("Ollama: http://localhost:11434")
print("Flask : http://localhost:8080")
print()
print("Open this in your browser:")
print()
print("    http://localhost:8080")
print()
print("=" * 60)
```
