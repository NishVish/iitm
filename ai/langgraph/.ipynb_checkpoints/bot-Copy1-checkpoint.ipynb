```python
from flask import Flask, request, jsonify
import requests
import threading
import json

from db_connection import get_connection


app = Flask(__name__)


# ============================================================
# CONFIG
# ============================================================

OLLAMA_URL = "http://127.0.0.1:11434/api/chat"
MODEL = "qwen2.5:7b"


# ============================================================
# SYSTEM PROMPT
# ============================================================

SYSTEM_PROMPT = """
You are Ishan Agent.

You have access to a MySQL database.

Your job is to understand what the user wants and, when database
information is required, generate SQL to retrieve that information.

IMPORTANT:

1. First understand the user's request.

2. If the question requires database information, generate SQL.

3. The database contains business/company information.

4. The main company table is:

company_data

Important columns include:

- id
- company_id
- company_name
- category
- subcategory
- address
- city
- state
- country
- website
- phone
- gst_number
- sales_person
- active_inactive
- entry_type
- total_staff
- created_at
- updated_at

5. When database information is required, respond ONLY with JSON:

{
    "action": "sql",
    "query": "SQL QUERY HERE"
}

6. When database information is NOT required, respond normally:

{
    "action": "answer",
    "response": "YOUR ANSWER"
}

7. NEVER invent database results.

8. Only generate SELECT queries.

9. Never generate INSERT, UPDATE, DELETE, DROP, ALTER, TRUNCATE,
CREATE or other destructive SQL.

10. If the user asks:

"total number of companies"

generate:

{
    "action": "sql",
    "query": "SELECT COUNT(*) AS total_companies FROM company_data"
}

11. After database results are provided to you, answer the user's
original question naturally.

You are Ishan Agent.

Never identify yourself as Qwen, ChatGPT, or Ollama.
"""


# ============================================================
# DATABASE
# ============================================================

def execute_sql(query):

    db = None
    cursor = None

    try:

        # Safety check
        query_clean = query.strip().lower()

        if not query_clean.startswith("select"):
            raise Exception(
                "Only SELECT queries are allowed."
            )

        # Block dangerous SQL keywords
        forbidden = [
            "insert ",
            "update ",
            "delete ",
            "drop ",
            "alter ",
            "truncate ",
            "create ",
            "replace ",
            "grant ",
            "revoke "
        ]

        for word in forbidden:

            if word in query_clean:

                raise Exception(
                    "Unsafe SQL query blocked."
                )

        db = get_connection()

        print("DATABASE CONNECTED:", db.is_connected())

        cursor = db.cursor(dictionary=True)

        print("\nSQL GENERATED BY QWEN:")
        print(query)

        cursor.execute(query)

        rows = cursor.fetchall()

        print("\nDATABASE RESULT:")
        print(rows)

        return rows

    except Exception as e:

        print("DATABASE ERROR:", str(e))

        raise

    finally:

        if cursor:
            cursor.close()

        if db:
            db.close()


# ============================================================
# CALL QWEN
# ============================================================

def ask_qwen(messages):

    response = requests.post(

        OLLAMA_URL,

        json={
            "model": MODEL,

            "messages": messages,

            "stream": False,

            "options": {
                "temperature": 0
            }
        },

        timeout=300
    )

    response.raise_for_status()

    result = response.json()

    return result["message"]["content"].strip()


# ============================================================
# AGENT
# ============================================================

@app.route("/agent", methods=["POST"])
def agent():

    data = request.get_json(silent=True)

    if not data or "message" not in data:

        return jsonify({
            "success": False,
            "error": "message is required"
        }), 400


    user_message = str(
        data["message"]
    ).strip()


    if not user_message:

        return jsonify({
            "success": False,
            "error": "message cannot be empty"
        }), 400


    print("\n")
    print("=" * 60)
    print("USER QUERY:")
    print(user_message)
    print("=" * 60)


    try:

        # ----------------------------------------------------
        # STEP 1
        # Ask Qwen what the user wants
        # ----------------------------------------------------

        messages = [

            {
                "role": "system",
                "content": SYSTEM_PROMPT
            },

            {
                "role": "user",
                "content": user_message
            }

        ]


        first_response = ask_qwen(messages)


        print("\nQWEN DECISION:")
        print(first_response)


        # ----------------------------------------------------
        # Parse Qwen response
        # ----------------------------------------------------

        try:

            decision = json.loads(
                first_response
            )

        except json.JSONDecodeError:

            return jsonify({

                "success": True,

                "response": first_response

            })


        action = decision.get(
            "action"
        )


        # ====================================================
        # NORMAL QUESTION
        # ====================================================

        if action == "answer":

            return jsonify({

                "success": True,

                "response":
                    decision.get(
                        "response",
                        ""
                    )

            })


        # ====================================================
        # DATABASE QUESTION
        # ====================================================

        if action == "sql":

            query = decision.get(
                "query"
            )


            if not query:

                return jsonify({

                    "success": False,

                    "error":
                        "Qwen requested SQL but did not provide a query."

                }), 500


            # ------------------------------------------------
            # STEP 2
            # Execute SQL
            # ------------------------------------------------

            database_result = execute_sql(
                query
            )


            # ------------------------------------------------
            # STEP 3
            # Give database result back to Qwen
            # ------------------------------------------------

            messages.append({

                "role": "assistant",

                "content":
                    first_response

            })


            messages.append({

                "role": "user",

                "content": f"""
The database executed your SQL query.

Database result:

{json.dumps(
    database_result,
    default=str
)}

Now answer the user's original question.

Do not mention SQL unless necessary.
Do not invent information.
"""

            })


            final_response = ask_qwen(
                messages
            )


            print("\nFINAL ANSWER:")
            print(final_response)

            print("=" * 60)


            return jsonify({

                "success": True,

                "response": final_response,

                "sql": query

            })


        # ====================================================
        # UNKNOWN ACTION
        # ====================================================

        return jsonify({

            "success": False,

            "error":
                "Unknown action returned by Qwen.",

            "qwen_response":
                first_response

        }), 500


    # ========================================================
    # ERRORS
    # ========================================================

    except requests.exceptions.ConnectionError:

        return jsonify({

            "success": False,

            "error":
                "Cannot connect to Ollama. Make sure Ollama is running."

        }), 500


    except requests.exceptions.Timeout:

        return jsonify({

            "success": False,

            "error":
                "Ollama request timed out."

        }), 504


    except Exception as e:

        print("\nERROR:")
        print(str(e))


        return jsonify({

            "success": False,

            "error": str(e)

        }), 500


# ============================================================
# START SERVER
# ============================================================

def run_server():

    app.run(

        host="127.0.0.1",

        port=1234,

        debug=False,

        use_reloader=False

    )


server_thread = threading.Thread(

    target=run_server,

    daemon=True

)


server_thread.start()


print()
print("=" * 60)
print("ISHAN AGENT STARTED")
print("=" * 60)
print()
print("Model  :", MODEL)
print("Ollama :", OLLAMA_URL)
print("Agent  : http://127.0.0.1:1234/agent")
print()
print("=" * 60)
```
