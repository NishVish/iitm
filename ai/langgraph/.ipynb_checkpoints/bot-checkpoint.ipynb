````python
# ============================================================
# ISHAN AGENT (Optimized)
# Qwen 2.5 7B + Ollama + MySQL
#
# FLOW:
#
# User
#   ↓
# Qwen understands request -> emits JSON decision
#   ↓
# If action == "sql": Python executes SQL against MySQL
#   ↓
# MySQL result is fed back to Qwen
#   ↓
# Qwen turns the raw result into a plain-English final answer
#   ↓
# Flask returns ONLY the final answer to the caller
#   (raw SQL / raw rows are available only if ?debug=1 is passed)
#
# Flask: http://127.0.0.1:1234/agent
# ============================================================

import json
import logging
import re
import threading
import time

import requests
from flask import Flask, request, jsonify

from db_connection import get_connection


# ============================================================
# CONFIG
# ============================================================

app = Flask(__name__)

OLLAMA_URL = "http://127.0.0.1:11434/api/chat"
MODEL = "qwen2.5:7b"

OLLAMA_TIMEOUT_SECONDS = 120          # was 300 — 7B model should respond fast; fail sooner
SQL_ROW_LIMIT = 500                    # hard ceiling injected into every SELECT that lacks one
SIMPLE_CACHE_TTL_SECONDS = 20          # short cache for identical, back-to-back questions

logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s [%(levelname)s] %(message)s",
)
log = logging.getLogger("ishan_agent")

# Reuse one TCP connection to Ollama instead of opening a new socket per call.
_session = requests.Session()

# Tiny in-memory cache: {user_message_lower: (timestamp, response_json)}
_cache_lock = threading.Lock()
_response_cache = {}


# ============================================================
# SYSTEM PROMPT
# ============================================================

SYSTEM_PROMPT = """
You are Ishan Agent.

You have access to a MySQL business database.

Your job is to understand the user's request.

If the user asks for information that must come from the
database, generate a SQL SELECT query.

Return ONLY valid JSON, with no markdown fences and no extra text.

For a database request:

{"action": "sql", "query": "SELECT ..."}

For a normal conversation request:

{"action": "answer", "response": "..."}

DATABASE:

Table: company_data

Columns:
id, company_id, database_name, company_name, category, subcategory,
address, city, pincode, state, country, website, phone, gst_number,
sales_person, active_inactive, created_at, updated_at, last_confirmed_at,
entry_type, pin, travel_segments, meet_profiles, meet_regions,
interested_states, branch_offices, total_staff, association_membership

RULES:

- Database questions require SQL.
- Only generate SELECT queries. Never INSERT/UPDATE/DELETE/DROP/ALTER/TRUNCATE.
- Never invent database results — only use what the database returns to you later.
- Use the actual table and column names.
- Return JSON only.

Example:

User: total number of companies in our database
Return: {"action": "sql", "query": "SELECT COUNT(*) AS total_companies FROM company_data"}
"""

FINAL_ANSWER_SYSTEM_PROMPT = """
You are Ishan Agent speaking directly to a business user.

You will be given the user's original question and the exact data
returned from the database for that question.

Answer in plain, natural English — a normal sentence, not JSON, not
markdown, not code. Do not mention SQL, JSON, tables, columns, or
the database. Do not invent any numbers or facts not present in the
data you were given. If the data is an empty list, say plainly that
no matching records were found.
"""

FORBIDDEN_PATTERN = re.compile(
    r"\b(insert|update|delete|drop|alter|truncate|create|replace|grant|revoke|"
    r"into|outfile|load_file|information_schema)\b",
    re.IGNORECASE,
)
MULTI_STATEMENT_PATTERN = re.compile(r";\s*\S")  # semicolon followed by more code = stacked query


# ============================================================
# CALL QWEN
# ============================================================

def ask_qwen(messages):
    response = _session.post(
        OLLAMA_URL,
        json={
            "model": MODEL,
            "messages": messages,
            "stream": False,
            "options": {"temperature": 0},
        },
        timeout=OLLAMA_TIMEOUT_SECONDS,
    )
    response.raise_for_status()
    return response.json()["message"]["content"].strip()


def extract_json(raw_text):
    """
    Robustly pull a JSON object out of a model response, even if the
    model wraps it in markdown fences or adds stray text around it.
    """
    text = raw_text.strip()

    # Strip ```json ... ``` or ``` ... ``` fences if present.
    fence_match = re.search(r"```(?:json)?\s*(.*?)\s*```", text, re.DOTALL)
    if fence_match:
        text = fence_match.group(1).strip()

    try:
        return json.loads(text)
    except json.JSONDecodeError:
        pass

    # Fallback: grab the first {...} block in the text.
    brace_match = re.search(r"\{.*\}", text, re.DOTALL)
    if brace_match:
        return json.loads(brace_match.group(0))

    raise ValueError(f"Could not parse JSON from model output: {raw_text!r}")


def unwrap_if_json(text):
    """
    Defensive net: if the model ignores instructions and still answers
    in {"action": "answer", "response": "..."} form (or similar), pull
    the human-readable string out instead of returning raw JSON to the
    caller.
    """
    stripped = text.strip()
    if not stripped.startswith("{"):
        return text

    try:
        parsed = extract_json(stripped)
    except (ValueError, json.JSONDecodeError):
        return text

    if isinstance(parsed, dict):
        for key in ("response", "answer", "message"):
            if key in parsed and isinstance(parsed[key], str):
                return parsed[key]

    return text


# ============================================================
# SQL SAFETY + EXECUTION
# ============================================================

def enforce_select_only(query):
    clean = query.strip().rstrip(";").strip()

    if not clean.lower().startswith("select"):
        raise ValueError("Only SELECT queries are allowed.")

    if MULTI_STATEMENT_PATTERN.search(clean):
        raise ValueError("Stacked/multiple statements are not allowed.")

    if FORBIDDEN_PATTERN.search(clean):
        raise ValueError("Unsafe SQL query blocked.")

    # Guard against accidentally scanning huge tables.
    if not re.search(r"\blimit\b", clean, re.IGNORECASE):
        clean = f"{clean} LIMIT {SQL_ROW_LIMIT}"

    return clean


def execute_sql(query):
    safe_query = enforce_select_only(query)

    log.info("EXECUTING SQL: %s", safe_query)

    db = None
    cursor = None
    try:
        db = get_connection()
        cursor = db.cursor(dictionary=True)
        cursor.execute(safe_query)
        rows = cursor.fetchall()
        log.info("ROWS RETURNED: %d", len(rows))
        log.info("QUERY RESULT: %s", json.dumps(rows, default=str, indent=2))
        return rows, safe_query
    finally:
        if cursor:
            cursor.close()
        if db:
            db.close()


# ============================================================
# AGENT CORE (separated from the Flask route for clarity/testability)
# ============================================================

def run_agent(user_message):
    messages = [
        {"role": "system", "content": SYSTEM_PROMPT},
        {"role": "user", "content": user_message},
    ]

    qwen_response = ask_qwen(messages)
    log.info("QWEN DECISION: %s", qwen_response)

    decision = extract_json(qwen_response)
    action = decision.get("action")

    if action == "answer":
        return {
            "response": decision.get("response", ""),
        }

    if action == "sql":
        query = decision.get("query")
        if not query:
            raise ValueError("Model returned action='sql' but no query.")

        rows, safe_query = execute_sql(query)

        # IMPORTANT: use a fresh conversation for this step, WITHOUT the
        # original system prompt. If we reuse the JSON-mandating system
        # prompt here, small models (Qwen 7B included) tend to keep
        # replying in JSON — which means the "answer" the caller sees
        # is literally the raw {"action": "sql", "query": "..."} blob,
        # not a real sentence. That was the original bug.
        result_prompt = f"""
USER'S ORIGINAL QUESTION:
{user_message}

DATABASE RESULT (JSON rows):
{json.dumps(rows, default=str)}
"""
        final_messages = [
            {"role": "system", "content": FINAL_ANSWER_SYSTEM_PROMPT},
            {"role": "user", "content": result_prompt},
        ]

        final_answer = ask_qwen(final_messages)
        final_answer = unwrap_if_json(final_answer)
        log.info("FINAL ANSWER: %s", final_answer)

        return {
            "response": final_answer,
            "sql": safe_query,
            "database_result": rows,
        }

    raise ValueError(f"Model returned an unknown action: {action!r}")


# ============================================================
# FLASK ROUTE
# ============================================================

@app.route("/agent", methods=["POST"])
def agent():
    data = request.get_json(silent=True)

    if not data or "message" not in data:
        return jsonify({"success": False, "error": "message is required"}), 400

    user_message = str(data["message"]).strip()
    if not user_message:
        return jsonify({"success": False, "error": "message cannot be empty"}), 400

    debug = request.args.get("debug", "0") == "1"
    cache_key = user_message.lower()

    log.info("=" * 60)
    log.info("USER QUERY: %s", user_message)

    # Serve identical, rapid-fire questions from cache to save an
    # Ollama + MySQL round trip.
    with _cache_lock:
        cached = _response_cache.get(cache_key)
        if cached and (time.time() - cached[0]) < SIMPLE_CACHE_TTL_SECONDS:
            log.info("CACHE HIT")
            result = cached[1]
            payload = {"success": True, "response": result["response"]}
            if debug:
                payload.update({k: v for k, v in result.items() if k != "response"})
            return jsonify(payload)

    try:
        result = run_agent(user_message)

        with _cache_lock:
            _response_cache[cache_key] = (time.time(), result)

        payload = {"success": True, "response": result["response"]}
        if debug:
            payload.update({k: v for k, v in result.items() if k != "response"})

        return jsonify(payload)

    except requests.exceptions.ConnectionError:
        log.error("Cannot connect to Ollama.")
        return jsonify({
            "success": False,
            "error": "Cannot connect to Ollama. Make sure Ollama is running.",
        }), 500

    except requests.exceptions.Timeout:
        log.error("Ollama request timed out.")
        return jsonify({"success": False, "error": "Ollama request timed out."}), 504

    except ValueError as e:
        # Expected/validation-type errors (bad SQL, bad JSON, blocked query, etc.)
        log.warning("AGENT VALIDATION ERROR: %s", e)
        return jsonify({"success": False, "error": str(e)}), 400

    except Exception as e:
        log.exception("AGENT ERROR")
        return jsonify({"success": False, "error": str(e)}), 500


# ============================================================
# START FLASK
# ============================================================

def run_server():
    app.run(host="127.0.0.1", port=1234, debug=False, use_reloader=False)


server_thread = threading.Thread(target=run_server, daemon=True)
server_thread.start()

log.info("=" * 60)
log.info("ISHAN AGENT STARTED")
log.info("MODEL  : %s", MODEL)
log.info("OLLAMA : %s", OLLAMA_URL)
log.info("AGENT  : http://127.0.0.1:1234/agent")
log.info("=" * 60)
````
