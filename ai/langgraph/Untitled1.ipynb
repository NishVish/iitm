```python
import requests

url = "http://localhost:11434/api/chat"

payload = {
    "model": "qwen2.5:7b",
    "messages": [
        {
            "role": "user",
            "content": "Hello, are you working?"
        }
    ],
    "stream": False
}

response = requests.post(url, json=payload)

print("STATUS:", response.status_code)
print("RESPONSE:")
print(response.text)
```
