```python
import matplotlib.pyplot as plt
import numpy as np

# Data
years = ["2024-2025", "2025-2026", "2026-2027"]

visitors = [2200, 4250, 4560]
exhibitors = [250, 262, 276]

visitor_growth = ["", "+93.2% YoY", "+7.3% YoY"]
exhibitor_growth = ["", "+4.8% YoY", "+5.3% YoY"]

x = np.arange(len(years))

# Create figure
fig, ax1 = plt.subplots(figsize=(10, 6))

# Bar width
width = 0.35

# Visitors bars
bars1 = ax1.bar(
    x - width/2,
    visitors,
    width,
    label="Trade Visitors",
    color="#1f77b4"
)

# Exhibitors bars (secondary axis)
ax2 = ax1.twinx()

bars2 = ax2.bar(
    x + width/2,
    exhibitors,
    width,
    label="Exhibitors",
    color="#ff7f0e"
)

# Labels
ax1.set_xlabel("Year")
ax1.set_ylabel("Trade Visitors", color="#1f77b4")
ax2.set_ylabel("Exhibitors", color="#ff7f0e")

ax1.set_xticks(x)
ax1.set_xticklabels(years)

ax1.set_title(
    "Trade Visitors & Exhibitor Growth (2024-2027)",
    fontsize=14,
    fontweight="bold"
)

# Add values on bars
for bar in bars1:
    height = bar.get_height()
    ax1.text(
        bar.get_x() + bar.get_width()/2,
        height + 100,
        f"{height:,}",
        ha="center",
        fontsize=10
    )

for bar in bars2:
    height = bar.get_height()
    ax2.text(
        bar.get_x() + bar.get_width()/2,
        height + 5,
        f"{height}",
        ha="center",
        fontsize=10
    )

# Add growth annotations
for i, text in enumerate(visitor_growth):
    if text:
        ax1.annotate(
            text,
            (x[i]-width/2, visitors[i]),
            xytext=(0, 25),
            textcoords="offset points",
            ha="center",
            color="green",
            fontweight="bold"
        )

for i, text in enumerate(exhibitor_growth):
    if text:
        ax2.annotate(
            text,
            (x[i]+width/2, exhibitors[i]),
            xytext=(0, 15),
            textcoords="offset points",
            ha="center",
            color="darkorange",
            fontweight="bold"
        )

# Legend
fig.legend(
    loc="upper left",
    bbox_to_anchor=(0.1, 0.9)
)

plt.tight_layout()
plt.show()
```
