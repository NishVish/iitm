```python
# Make sure you are in the correct Jupyter kernel (venv_iitm)

import pandas as pd          # For Excel and data handling
import requests              # For HTTP requests if needed
from bs4 import BeautifulSoup # For HTML parsing
from datetime import datetime # For handling dates
import time                  # For delays or timing
import openpyxl              # Excel engine for .xlsx files

# Path to your Excel file
file_path = "companies.xlsx"

# Read the 'main' sheet
df = pd.read_excel(file_path, sheet_name="main", engine="openpyxl")

# Fill empty strings for text columns, 0 for numeric columns
for col in df.columns:
    if df[col].dtype == 'object':
        df[col].fillna('', inplace=True)
    else:
        df[col].fillna(0, inplace=True)

# Preview the dataframe
df.head()
```
