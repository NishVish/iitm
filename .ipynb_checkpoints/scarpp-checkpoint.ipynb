```python
import requests
from bs4 import BeautifulSoup
import re

URL = "https://iitmindia.com/trade-visitor2/"

def clean_text(text):
    if not text:
        return ""
    return re.sub(r"\s+", " ", text.strip())

def scrape_iitm_events():
    response = requests.get(URL, timeout=15)
    response.raise_for_status()

    soup = BeautifulSoup(response.text, "html.parser")

    events = []

    # Each event is inside an elementor column
    columns = soup.find_all("div", class_="elementor-column")

    for col in columns:
        texts = [clean_text(t.get_text()) for t in col.find_all(["h5", "p"])]

        if not texts:
            continue

        event = {
            "name": None,
            "dates": None,
            "location": None,
            "time": None,
            "status": "Registration Open"
        }

        for t in texts:
            if "IITM" in t and not event["name"]:
                event["name"] = t

            elif "AM" in t or "PM" in t:
                event["time"] = t

            elif any(month in t for month in [
                "Jan", "Feb", "March", "April", "May", "June",
                "July", "Aug", "Sep", "Oct", "Nov", "Dec"
            ]) or "TBA" in t or "POSTPONED" in t:
                event["dates"] = t

            elif "Convention" in t or "Ground" in t or "Hall" in t or "Stadium" in t:
                event["location"] = t

        # Status detection
        full_text = " ".join(texts).upper()

        if "SUCCESSFULLY COMPLETED" in full_text:
            event["status"] = "Completed"
        elif "POSTPONED" in full_text:
            event["status"] = "Postponed"
        elif "TBA" in full_text and not event["dates"]:
            event["status"] = "TBA"

        if event["name"]:
            events.append(event)

    return events


if __name__ == "__main__":
    data = scrape_iitm_events()
    for e in data:
        print(e)
```

```python
import requests
from bs4 import BeautifulSoup
import re

URL = "https://iitmindia.com/trade-visitor2/"

def clean_text(text):
    if not text:
        return ""
    return re.sub(r"\s+", " ", text.strip())

def is_visible(tag):
    # If the element has inline style display:none, skip it
    style = tag.get("style", "")
    if "display:none" in style.replace(" ", "").lower():
        return False
    return True

def scrape_visible_events():
    response = requests.get(URL, timeout=15)
    response.raise_for_status()

    soup = BeautifulSoup(response.text, "html.parser")
    events = []

    # Each event is inside an elementor column
    columns = soup.find_all("div", class_="elementor-column")

    for col in columns:
        # Skip hidden columns
        if not is_visible(col):
            continue

        texts = [clean_text(t.get_text()) for t in col.find_all(["h5", "p"]) if is_visible(t)]

        if not texts:
            continue

        event = {
            "name": None,
            "dates": None,
            "location": None,
            "time": None,
            "status": "Registration Open"
        }

        for t in texts:
            if "IITM" in t and not event["name"]:
                event["name"] = t

            elif "AM" in t or "PM" in t:
                event["time"] = t

            elif any(month in t for month in [
                "Jan", "Feb", "March", "April", "May", "June",
                "July", "Aug", "Sep", "Oct", "Nov", "Dec"
            ]) or "TBA" in t or "POSTPONED" in t:
                event["dates"] = t

            elif "Convention" in t or "Ground" in t or "Hall" in t or "Stadium" in t:
                event["location"] = t

        # Status detection
        full_text = " ".join(texts).upper()
        if "SUCCESSFULLY COMPLETED" in full_text:
            event["status"] = "Completed"
        elif "POSTPONED" in full_text:
            event["status"] = "Postponed"
        elif "TBA" in full_text and not event["dates"]:
            event["status"] = "TBA"

        if event["name"]:
            events.append(event)

    return events

if __name__ == "__main__":
    visible_events = scrape_visible_events()
    for e in visible_events:
        print(e)
```

```python
import requests
from bs4 import BeautifulSoup
from urllib.parse import urljoin

BASE_URL = "https://iitmindia.com"

def get_trade_visitor_url():
    r = requests.get(BASE_URL, timeout=15)
    r.raise_for_status()
    soup = BeautifulSoup(r.text, "html.parser")
    link = soup.find("a", string=lambda s: s and "Trade Visitor" in s)
    if not link:
        raise Exception("Trade Visitor link not found on homepage")
    href = link.get("href")
    return urljoin(BASE_URL, href)

def is_visible(element):
    if element is None:
        return False
    style = element.get("style", "")
    if "display:none" in style.replace(" ", "").lower():
        return False
    hidden_classes = {"hidden", "d-none", "elementor-hidden"}
    classes = set(element.get("class", []))
    if classes.intersection(hidden_classes):
        return False
    if element.get("aria-hidden") == "true":
        return False
    return True

def get_visible_text(tag):
    texts = []
    for child in tag.descendants:
        if child.name:
            if not is_visible(child):
                continue
        if isinstance(child, str):
            t = child.strip()
            if t:
                texts.append(t)
    return texts

def scrape_and_group_events():
    trade_url = get_trade_visitor_url()
    r = requests.get(trade_url, timeout=15)
    r.raise_for_status()
    soup = BeautifulSoup(r.text, "html.parser")

    events = []
    seen = set()

    # Only target the main container with event listings
    main_container = soup.find("div", class_="elementor-widget-wrap")
    if not main_container:
        return events

    sections = main_container.find_all("section", class_="elementor-inner-section")
    for section in sections:
        if not is_visible(section):
            continue
        cols = section.find_all("div", class_="elementor-column")
        for col in cols:
            if not is_visible(col):
                continue
            texts = get_visible_text(col)
            if not texts:
                continue
            if texts[0].upper() in {"HOME", "ESCALATE YOUR BRAND VISIBILITY WITH \u200b"}:
                continue

            event_name = texts[0].strip()
            # skip duplicates or previous years
            if event_name in seen:
                continue
            seen.add(event_name)

            event_data = {
                "name": event_name,
                "dates": None,
                "location": None,
                "time": None,
                "status_text": None
            }

            status_lines = []
            for line in texts[1:]:
                line_upper = line.upper()
                if "AM" in line or "PM" in line:
                    event_data["time"] = line
                elif "POSTPONED" in line_upper or "SUCCESSFULLY COMPLETED" in line_upper:
                    status_lines.append(line)
                elif any(month in line for month in [
                    "Jan","Feb","Mar","April","May","June",
                    "July","Aug","Sept","Oct","Nov","Dec"
                ]) or "TBA" in line:
                    event_data["dates"] = line
                elif any(word in line for word in [
                    "Hall","Ground","Center","Centre","Stadium","Complex","Gate"
                ]):
                    event_data["location"] = line
                else:
                    status_lines.append(line)

            if status_lines:
                event_data["status_text"] = "\n".join(status_lines)

            events.append(event_data)

    return events

if __name__ == "__main__":
    grouped_events = scrape_and_group_events()
    for event in grouped_events:
        print("-" * 40)
        print(event)
```

```python
import requests
from bs4 import BeautifulSoup
from urllib.parse import urljoin

BASE_URL = "https://iitmindia.com"

# Step 1: Get homepage and find "Trade Visitor" link
r = requests.get(BASE_URL, timeout=15)
r.raise_for_status()
soup = BeautifulSoup(r.text, "html.parser")

link = soup.find("a", string=lambda s: s and "Trade Visitor" in s)
if not link:
    print("Trade Visitor link not found.")
    exit()

trade_url = urljoin(BASE_URL, link.get("href"))
print("Navigating to:", trade_url)

# Step 2: Get the Trade Visitor page
r2 = requests.get(trade_url, timeout=15)
r2.raise_for_status()
soup2 = BeautifulSoup(r2.text, "html.parser")

# Step 3: Extract all event-like entries
# Assuming each event is inside a column container
event_containers = soup2.find_all("div", class_="elementor-column")

def is_visible(element):
    """Check if an element is visible"""
    if element is None:
        return False
    style = element.get("style", "")
    if "display:none" in style.replace(" ", "").lower():
        return False
    hidden_classes = {"hidden", "d-none", "elementor-hidden"}
    classes = set(element.get("class", []))
    if classes.intersection(hidden_classes):
        return False
    if element.get("aria-hidden") == "true":
        return False
    return True

def get_text_list(tag):
    """Get visible text as a list of lines"""
    texts = []
    for child in tag.descendants:
        if child.name and not is_visible(child):
            continue
        if isinstance(child, str):
            t = child.strip()
            if t:
                texts.append(t)
    return texts

seen = set()
for col in event_containers:
    if not is_visible(col):
        continue
    texts = get_text_list(col)
    if not texts:
        continue
    # Filter out irrelevant sections like "HOME" or promotional text
    if texts[0].upper() in {"HOME", "ESCALATE YOUR BRAND VISIBILITY WITH \u200b"}:
        continue
    event_name = texts[0]
    if event_name in seen:
        continue
    seen.add(event_name)

    # Print the entry
    print("-" * 50)
    for line in texts:
        print(line)
```

```python
import requests
from bs4 import BeautifulSoup
from urllib.parse import urljoin

BASE_URL = "https://iitmindia.com"

# Step 1: Get homepage and find "Trade Visitor" link
r = requests.get(BASE_URL, timeout=15)
r.raise_for_status()
soup = BeautifulSoup(r.text, "html.parser")

link = soup.find("a", string=lambda s: s and "Trade Visitor" in s)
if not link:
    print("Trade Visitor link not found.")
    exit()

trade_url = urljoin(BASE_URL, link.get("href"))

# Step 2: Get the Trade Visitor page
r2 = requests.get(trade_url, timeout=15)
r2.raise_for_status()
soup2 = BeautifulSoup(r2.text, "html.parser")

# Step 3: Extract event containers
event_containers = soup2.find_all("div", class_="elementor-column")

def is_visible(element):
    if element is None:
        return False
    style = element.get("style", "")
    if "display:none" in style.replace(" ", "").lower():
        return False
    hidden_classes = {"hidden", "d-none", "elementor-hidden"}
    classes = set(element.get("class", []))
    if classes.intersection(hidden_classes):
        return False
    if element.get("aria-hidden") == "true":
        return False
    return True

def get_text_list(tag):
    texts = []
    for child in tag.descendants:
        if child.name and not is_visible(child):
            continue
        if isinstance(child, str):
            t = child.strip()
            if t:
                texts.append(t)
    return texts

seen = set()
for col in event_containers:
    if not is_visible(col):
        continue
    texts = get_text_list(col)
    if not texts:
        continue
    if texts[0].upper() in {"HOME", "ESCALATE YOUR BRAND VISIBILITY WITH \u200b"}:
        continue
    event_name = texts[0]
    if event_name in seen:
        continue
    seen.add(event_name)

    # Initialize fields
    name = event_name
    dates = None
    location = None
    time = None
    registration = None
    status = None

    for line in texts[1:]:
        line_upper = line.upper()
        if "AM" in line or "PM" in line:
            time = line
        elif "POSTPONED" in line_upper or "SUCCESSFULLY COMPLETED" in line_upper:
            status = line
        elif any(month in line for month in [
            "Jan","Feb","Mar","April","May","June",
            "July","Aug","Sept","Sep","Oct","Nov","Dec"
        ]) or "TBA" in line:
            dates = line
        elif any(word in line for word in [
            "Hall","Ground","Center","Centre","Stadium","Complex","Gate"
        ]):
            location = line
        elif "Trade Visitor" in line:
            registration = line
        else:
            if not location:
                location = line
            else:
                status = line

    # Print in requested format
    print("-" * 50)
    print(name)
    if status == "POSTPONED":
        print(status)
    if dates:
        print(dates)
    if location and status != "POSTPONED":
        print(location)
    if time:
        print(f"Time {time}")
    if registration:
        print(registration)
    if status and status not in {None, "POSTPONED"}:
        print(status)
```
